```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from statsmodels.tsa.statespace.sarimax import SARIMAX
```

```python
dates = pd.date_range(
    start="2022-01-01",
    periods=48,
    freq="MS"
)

np.random.seed(42)
sales = (
    100
    + np.arange(48) * 2
    + 15 * np.sin(2 * np.pi * np.arange(48) / 12)
    + np.random.normal(0, 5, 48)
)

df = pd.DataFrame({
    "sales": sales
}, index=dates)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df["sales"])
plt.title("Evolution des ventes mensuelles")
plt.xlabel("Date")
plt.ylabel("Ventes")

plt.show()
```

```python
model = SARIMAX(
    df["sales"],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12)
)
results = model.fit()
print(results.summary())
```

```python
forecast = results.get_forecast(steps=12)

predictions = forecast.predicted_mean

confidence = forecast.conf_int()

plt.figure(figsize=(12, 5))

plt.plot(df["sales"], label="Ventes réelles")

plt.plot(
    predictions,
    label="Prévisions SARIMAX",
    color="red"
)

plt.fill_between(
    confidence.index,
    confidence.iloc[:, 0],
    confidence.iloc[:, 1],
    alpha=0.2
)

plt.legend()
plt.title("Prévision des ventes sur 12 mois")
plt.xlabel("Date")
plt.ylabel("Ventes")
plt.grid(True)
plt.show()
```
